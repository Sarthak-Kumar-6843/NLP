```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
```

```python
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt) : 
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new
    
df['text'] = df['text'].apply(remove_numbers)  
```

```python
def remove_urls(txt):
    new = ""
    for word in txt.split():
        if not word.startswith(('http://', 'https://', 'www.')):
            new = new + word + " "
    return new.strip()

df['text'] = df['text'].apply(remove_urls)
```

```python
def remove_html_tags(txt):
    new = ""
    inside_tag = False

    for i in txt:
        if i == '<':
            inside_tag = True
        elif i == '>':
            inside_tag = False
        elif not inside_tag:
            new = new + i

    return new

df['text'] = df['text'].apply(remove_html_tags)
```

```python
def remove_emoji(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new        

df['text'] = df['text'].apply(remove_emoji)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
            
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```
